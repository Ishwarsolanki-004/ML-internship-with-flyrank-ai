## My Lane: Refresh / Content Opportunity Scoring

I chose the **Refresh / Content Opportunity Scoring** lane because the goal is to help content editors decide **which content pages should be reviewed first**.

This is a **ranking/scoring** problem rather than a simple classification problem. Instead of predicting only "refresh" or "do not refresh", the system produces a priority score so that editors can review the highest-value pages first.

The output supports a real editorial decision by creating a ranked review queue, helping teams use their limited review time more effectively.


```python
ml_task = "Ranking / Scoring"

print("ML Task Type:", ml_task)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric is **Precision@50**.

Editors usually have time to review only a limited number of pages. Precision@50 measures how many of the top-ranked pages are genuinely useful refresh opportunities.

A higher Precision@50 means the ranking is more valuable for editorial decision-making.


```python
success_metric = "Precision@50"

print("Success Metric:", success_metric)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple rule such as "refresh every page with low traffic" ignores many important signals.

Content performance depends on multiple factors including impressions, clicks, CTR, average position, engagement, content age, and user behaviour. These signals interact in complex ways and may vary across clients and content types.

Machine learning can combine these signals into a single priority score, producing a more useful ranked review queue than a fixed threshold. The model is intended as a **decision-support tool**, helping editors focus their effort where it is most likely to have an impact.


```python
important_features = [
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "engagement_rate",
    "content_age_days"
]

print("Example signals considered for scoring:")

for feature in important_features:
    print("-", feature)
```
